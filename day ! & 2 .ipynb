# Type casting

### convert one data type to another data types

```python
# integer type casting: int()

v1 = int(2.8)  # float into int
v2 = int("100") # str to int
v3 = int(True) # boolean to int 
v4 = int(False)
v5 = int("0110", 2) #binary to int
v6 = int("A", 16)  #hexadecimal to int

print (v1, type(v1))
print (v2, type(v2))
print (v3, type(v3))
print (v4, type(v4))
print (v5, type(v5))
print (v6, type(v6))
```

```python
# float type casting : float()

v1 = float (10) # int to float
v2 = float ("10.8") # str to float
v3 = float(True) # boolean to float

print (v1, type(v1))
print (v2, type(v2))
print (v3, type(v3))
```

```python
# string type casting : str()

v1 = str(10) #int to str
v2 = str(10.8) #float to str
v3 = str(True)

print (v1, type(v1))
print (v2, type(v2))
print (v3, type(v3))
```
