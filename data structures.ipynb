##### Addition, Deletion, and updation in tuple

```python
Movies = ("anand", "godavari", "dilbechura", "happy", "orange", "MS Dhoni - The untold story")

# insertion
Movies = list(Movies)
Movies.append("fidaa")
Movies = tuple(Movies)
print(Movies)

# delete
Movies = list(Movies)
Movies.remove("dilbechura")
Movies = tuple(Movies)
print(Movies)

# update
Movies = list(Movies)
Movies[2] = "dangal"
Movies = tuple(Movies)
print(Movies)
```

#### set operations

```python
# containement operation: check if element exist in the set or not

fest = {"pongal", "dussera", "ganesh chaturthi", "diwali","ugadi"}
if "holi" in fest:
    print("happy holi")
else:
    print("enjoy")

if "holi" not in fest:
    print("happy holi")
```

```python
# set comparision operator

s1={1,2,3,4,5}
s2={1,2,3,4}

if s1 == s2:
    print("yes")
else:
    print("no")

if s1 != s2:
    print("yes")
else:
    print("no")
    
# superset operations
if s1 >= s2:
    print("yes")
else:
    print("no")

if s1 <= s2:
    print("yes")
else:
    print("no")

# proper set operations
if s1 > s2:
    print("yes")
else:
    print("no")

if s1 < s2:
    print("yes")
else:
    print("no")
```
