### OOP: Object Oriented Programming

```python
# procedure oriented language: the language in which everthing is design with the help of function.
# ex:  c language.

# Object oriented programming language:
# Which use the class & object concept also.
# ex: c++, java, python

# java is pure oops language. but c++ and python is not pure
```

```python
# class : its a blueprint and structured, it is a collection of objects.
# object : instance of the class. real things
```

```python
# example - 1 :  structure of class and object:

#created the class
class car:
    pass
    
obj = car()  # created the object
```

```python
# example - 2
# class vairable = the variable which directly defined inside the class.
#class variable value can be access by any project.

class car:              # class name
    country = "India"   # class variable
    obj = car()         # created a object
print(obj.country) 
print(car.country)     # can access by class variable name also

car.country = "japan"  # update class variable with class name
print(car.country)

obj.country = "germany"  # we can update the class variable using class name
print(car.country)
print(obj.country)       #this obj will create local variable
```

```python
# example - 3 : constructor function
# 1. it is defined by __init__(self)
# 2. it automatically executes when we create the object.
# 3. it is used to initiallize the object instance variable.

# self variable is pointing to current object
# self is keeping the record of object which is currently executing

class car:
    country = "india"
    def __init__(self):
        print("i am constructor function")
obj=car()
    
```

```python
#example - 2

class Google:
    country = "india"
    def __init__(self, n, s, d):
        self.name = n   # instance variable
        self.salary = s # instance variable
        self.desg = d   # instance variable
obj = Google("Niranjan", 1500000, "DA")

print(obj.name,obj.salary,obj.desg)
```
