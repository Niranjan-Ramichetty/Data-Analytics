#### Encapsulation

```python
# Binding(variables) the data + methods together
# Benefits of the encapsulation is mainly security (unauthorized access is not allowed)

# Type of attributes:
# 1. public: accessable from any where - inside or outside the class.
# how to define :  public intance variable are defiend directly.

# 2. protected: accessable from only inside the class and subclasses.
# how to define : protected variable use the _ (single underscore)
# it is not supposed to access outside the class although it can be access 

# 3. private: accessable only inside the class.
# how to define : private variable use the __ (double underscore)
```

```python
class marvels:
    def __init__(self, name, power, weapon):
        self.name = name  # public
        self._power = power  # protected
        self.__weapon = weapon  # private
    def display(self):
        print(self.name, self._power, self.__weapon)
obj = marvels("Thor", "Thunder", "Hammer")
obj.display()
print(obj.name)    # it is public variable, it is used in outside the class
# print(obj.__weapon)  - it is private variable so it cannot used in outside
print(obj._power)  # it is protected variable - it is used in subclasses also(here,obj = subclass)
                   # but it is not supposed to use outside the class
# print(obj.__weapon)  - it is private variable so it cannot used in outside
print(obj._marvels__weapon)  # indirectly, it can be used by using class name in front of variable name
```

### Polymorphism:

##### 1. function overloading 

```python
# code to overload to function in c++ or java
# c++, java type of polymorphism is not applicable in python direclty.

def a(var1):
    print(var1)
def a(var1, var2):
    print(var1, var2)

# a(10)
a(10, 20)
```

```python
from multipledispatch import dispatch

@dispatch(int)
def a(var1):
    print(var1)
    
@dispatch(int, int)
def a(var1, var2):
    print(var1, var2)
@dispatch(int,int,int)
def a(var1,var2,var3):
    print(var1,var2,var3)
a(10)
a(10, 20)
a()
```

###### overloading fucntion using "Default parameters"

```python
def a (var1, var2 = None):
    if var2 is not None:
        print(var1, var2)
    else:
        print(var1)
a(10)
a(10, 20)
a("RE")
```

##### 2. Method overriding

```python
''' method overiding is defined as if same name exists in the parent and child class,
 then child class function will override the parent class fucntion'''

class father:
    def car(self):
        print("swift")
class me(father):
    def car(self):
         print("THAR")
class sis(father):
    def car(self):
        print("polo")
obj = me()
obj.car()
obj = sis()
obj.car()
```

##### Operator overloading

```python
# def: taking multiple tasks from the same operator
```

##### example: arthmetic operator overloading  (+,-,*,/,%,**,//)

```python
# Target: overload the arthmatic operator for user defined objects.

class dosa:
    def __init__(self, variety, price):
        self.variety = variety
        self.price = price
    # add fucntion(): is helping to overload the + operator
    def __add__(self, obj2):
        return self.price + obj2.price
    def __sub__(self, obj2):
        return self.price - obj2.price
    def __mul__(self, obj2):
        return self.price * obj2.price
    def __truediv__(self, obj2):
        return self.price / obj2.price
    def __pow__(self, obj2):
        return self.price ** obj2.price
    def __floordiv__(self, obj2):
        return self.price // obj2.price
        
d1 = dosa("egg", 50)
d2 = dosa("double", 8)

total = d1 + d2
print(total)

print(d2 - d1)
print(d1 * d2)
print(d1 / d2)
print(d1 ** d2)
print(d1 // d2)
```
