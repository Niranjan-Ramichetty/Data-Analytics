### bitwise opeators (&,|,^,~,<<,>>)

```python
# bitwise operators works on binary values
# bitwise operators are faster than arthmetic operators
# bitwise operators are perform many other task very easily as compare to other operator
```

```python
# & => AND, / => OR, ^ => XOR
'''
a b ab
0 0 0
0 1 0        AND
1 0 0 
1 1 1
'''
'''
a b ab
0 0 0
0 1 1        OR
1 0 1 
1 1 1
'''
'''
a b ab
0 0 0
0 1 1        XOR
1 0 1 
1 1 0
'''

a = 10
b = 3
c = 10
d = 3
e = 10
f = 3

print (a & b) #binary function 0010 =2
print (c | d) #binary function 1011 =11
print (e ^ f) #binary function 1001 =9
```

```python
a = 6
b = 10
c = 11

print (a & b & c) #binary function 0010 =2
print (a | b | c)
print (a ^ b ^ c)
```

```python
a = 6
b = 10
c = 14
d = 5
e = 8
f = 11

print (a ^ b ^ c ^ d ^ e ^ f)
```

```python
# NOT (~)

val = 10
print (~val) # not will convert plus into minus or minus to plus with -1
```

```python
val = -1000
print (~val)
```

```python
# left shift (<<)

# def: Insert the 0 value at the LSB(least significant bit)
'''formula : var * 2^bits '''

var = 30
print (var<<2) 
# it is doubles the number(*2^1)

var1 = 10
print(var1<<2)

var2 = 2
print(var2 << 80)
# it is doubles the number(*2^2)
```

```python
#  right shift (>>)

#right shift by 1 bit is making the value half
'''formula : [var / 2^bits] '''

var = 10
print (var>>1)
print (var>>2)
```

### Assignments operators(+=,-=,*=,/=,%=,&=)

```python
x = 10
x += 25
print (x) # x=x+25
```

```python
x = 10
x -= 25
print (x) # x=x-25
```

```python
x = 10
x *= 25
print (x) # x=x*25
```

```python
x = 10
x /= 25
print (x) # x=x/25
```

```python
x = 10
x %= 25
print (x) # x=x%25
```

```python
x = 10
x &= 25
print (x) # x=x&25
```

```python
x = 10
x |= 25
print (x) # x=x|25
```

```python
x = 10
x //= 3
print (x) # x=x//25
```
