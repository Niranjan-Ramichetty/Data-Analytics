## output in python

```python
# different types of output in python:

1.concatenation using commas : concatenate the variable and strings
2.string concatenation using + : this only works with strings
3.str.format() method :
4.f - strings
5.escape sequences
6.end and sep parameters
    
```

### 1. concatenation using commas

```python
south_hero_name = "AA"
Age = 44
print("Hero: ",south_hero_name, "age: " ,Age)
```

## string concatenation using + 

```python
heroine ="sam"
age= 35
print("name : " + heroine + "  age: "+ str(age))
```

## str.format() method :

```python
hero = "AA"
villan = "rana dagubati"
age = 45
print("hero:{2} name:{1} age:{0} ".format(hero,villan,age))
print("hero:{a} name:{b} age:{c} ".format(a = hero, b = villan, c = age))
```

### 4.f - strings

```python
name = "vaibhav suryavanshi"
age = 15

print(f"name:@{name},age:{age}")
```

```python
print(f"name:@{name}")
print(f"age:{age}")
```

## escape sequences

```python
print(f"name:@{name}\nage:{age}")  # for new line
```

```python
print(f"name:@{name}\t age:{age}")   # for tab space
```

## end and sep parameters

```python
name = "vaibhav suryavanshi"
age = 15

print(f"name:{name}", end = " |=\n=\t " )
print(f"age:{age}")
```

```python
# sep: it is a print function parameter,which helped to separate the values with specified symbol
```

```python
name = "vaibhav suryavanshi"
age = 15
print(name,age, sep ="%")
```
