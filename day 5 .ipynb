# input in python

```python
val = input("Enter a number")
```

```python
print (val)
print (type(val))
```

```python
val = int (input("Enter a number"))
```

```python
print (val)
print (type(val))
```

```python
val = float (input("Enter a number"))
```

```python
print (val)
print (type(val))
```

### multiple input

```python
x,y,z = input("enter 3 names").split()
```

```python
print (x,
       y,
       z)
```

```python
x,y,z = input("enter 3 names").split(',')
```

```python
print (x,
       y,
       z)
```

```python
x,y,z = input("enter 3 names").split(':')
```

```python
print (x,
       y,
       z)
```

```python
x,y,z = map (int, input("enter 3 names").split())
```

```python
print (x,y,z)
print (type(x))
```

```python
x,y,z = map (float, input("enter 3 names").split())
```

```python
print (x,y,z)
print (type(x))
```
