## operation on strings

#### 1. concatenation (+)



```python
# this operation is used to combine two or more string

a = "chicken"
b = "roti"
c = "buttermilk"

res = a +" "+"with "+ b +" drink "+" "+ c
print(res)
```

#### 2. repetition (*)

```python
# this operation will repeat a same string multiple times

print("hi "*10)
```

#### 3. membership (in, not in)

```python
# this operation will be used for checks for the substring inside the actual string

hero = "allu arjun"

# (in)
print("arjun" in hero)           # it is a part of string         
print("Arjun" in hero)
print("arj" in hero)
# (not in)
print("Arj" not in hero)
print("arj" not in hero)
```

#### 4. indexing

```python
heroine = "samantha"

print(heroine[0])
print(heroine[-1])
print(heroine[0:3])  # (start:end:step)
print(heroine[0:6:2])
print(heroine[::])   # (by default end = last of the string)
```

```python
print(heroine[::-1])   # it will reverse the string
```

```python
print(heroine[-8:-5])  #-5-(+1) = -6
```

```python
print(heroine[-2:-5:-1]) 
```

```python
print(heroine[-7:])
```

##### 5. slicing

```python
breakfast = "idli with chutney and sambar"

print(breakfast[0:4])
print(breakfast[10:17])       #just like range ( str[start:end:step] )
print(breakfast[22:33])
print(breakfast[:])
```
