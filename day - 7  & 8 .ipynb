```python
# example - 1   # beginner

a = 10
b = 11
if a%2 == 0: # 10%2 => 0
    print (a)

#example - 2
if b%2 == 0: # 11%2 => 1  (cant give any output due to reminder is 1)
    print (b)
else :    
    print("odd")  # there is no output then we print "odd"
```

```python
# example - 3 # elif condition: it is used to write one more condition inside the else part

biryani = input("famous biryani in hyderabad: ")

if biryani == 'mehfil biryani':
    print("19th century")
elif biryani == "shahgosh biryani":
    print("best of mutton biryani")
elif biryani == "pista house":
    print ("tasty and expensive")
else:
    print ("PG biryani")
```

## match case statement

```python
# match case statement is used to replace multiple if - else condition.
# match case is similar to switch case in c and c++.
# match case are more advance then the switch case.
# match case is having advance features like handling of data structure and conditions.
```

```python
# example - 1 : structure of match case

num =int( input("enter a number"))

match num:
    case 10:
        print("this is ")
    case 20:
        print("niranjan")
    case _:
        print("sai hi")
        
```

```python
# example - 2 : match case with data structure

data = [10,20,30]

match data:
    case [1,2,3]:
        print("hello world")
    case [30,20,10]:
        print("this is my 1st program")
    case [10,20,30]:
        print("this is correct statement")
    case _:
        print("nothing will match")
        
        
```

```python
# example - 3 : match case using conditi

num = int(input("enter "))

match num:
    case x if x > 0:
        print("it is positive number")
    case num if  (num > -450):
        print("it is negative number")
    case _:
        print("it is zero")
```

```python
# interview question

x = 15*10-11+100/2

if x >3:
    print("A")
    if x%3 == 0 and x:
        print("B")
    elif x == 15:
        print("D")
    else:
        pass
else:
    print("E")
```

# while loop

```python
# example - 1 : structure of while loop

i = 0
while i <= 10:
    i += 1         #increment 
    print(i)
    

```

```python
# example - 2 : break statement
i = 1
while i <= 10:
    print(i)
    if i == 5:
        break
    i += 1
```

```python
# example - 3 : continue statement
# continue statement take the control to the start of the loop
# after executing continue statement, further next statement will skip

i = 1
while i <= 10:
    i += 1
    if i == 5 or i == 8:
        continue 
    print(i)
```

## for loop

```python
# example-1: print all the elements of a list using for loop

animals = ["simham", 'puli', 'barray', 'kothi']

for ele in animals:
    print(ele)
```

```python
for i in animals:
    print(i)
```

```python
#example-2 : accessing the characters from a string using for loop

str1="gurram"

for i in str1:
    print(i)
```

```python
# example-3 : for loop with range()function
# syntax : range(start,end,step)
# by default the start value is 0.
# start = 0
# stop/end = -1
# step = 1

for i in range(10):
    print(i)
```

```python
# start = 2
# stop/end = 5
# step = 1

for i in range(2,5,1):
    print(i)
```

```python
for i in range(1,10,3):
    print(i)
```

```python
# example: print all the odd numbers in between 10 and 50.

for i in range(11,50,2):
    print(i)
```

```python
# example-4: break statement in for loop
# break statement will bring the control out of the loop.

for num in range(1,6):
    if num == 3:
        break
    print(num)
```

```python
for num in range(1,6):
    print(num)
    if num == 3:
        break
```

```python
# example-5: continue statement in for loop
# continue statement take the control to the start of the loop.
# after executing continue statement, further next statement will skip.

for num in range(1,6):
    if num == 3:
        continue
    print(num)
```

```python
# example-6: else statement with for loop
# else block will execute only if loop executed successfully.

for i in range(10):
    print(i)
    if i == 6:
        break
    
else:
    print("execute only if loop completed successfully")

# if we are coming out the loop using break statement then else block will not execute.
```

```python
# example-7: nested loop

for i in range(1,4):    # outer loop =3
    for j in range(1,3):   # inner loop =2
        print("hi")        # 3*2=6

# 1. for every value of outer loop, inner loop will execute completely.
# 2. the body part of inner loop will execute n=3 * m=2 times.
```
